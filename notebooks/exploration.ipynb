```python
# ============================================
# 14. DÉTECTION DES VALEURS EXTRÊMES
# ============================================

# On utilise ici les quartiles pour repérer les valeurs
# qui peuvent être considérées comme extrêmes.

Q1 = df["SalePrice"].quantile(0.25)
Q3 = df["SalePrice"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["SalePrice"] < lower_bound) |
    (df["SalePrice"] > upper_bound)
]

print("Q1 :", Q1)
print("Q3 :", Q3)
print("IQR :", IQR)

print("\nLimite inférieure :", lower_bound)
print("Limite supérieure :", upper_bound)

print("\nNombre de valeurs extrêmes :", len(outliers))
```

```python
# ============================================
# 31. FEATURE ENGINEERING — SURFACE TOTALE
# ============================================

clean_df["TotalSF"] = (
    clean_df["TotalBsmtSF"]
    + clean_df["1stFlrSF"]
    + clean_df["2ndFlrSF"]
)

print(clean_df[[
    "TotalBsmtSF",
    "1stFlrSF",
    "2ndFlrSF",
    "TotalSF"
]].head())
```

```python
# ============================================
# 32. FEATURE ENGINEERING — TOTAL BATHROOMS
# ============================================

clean_df["TotalBathrooms"] = (
    clean_df["FullBath"]
    + 0.5 * clean_df["HalfBath"]
    + clean_df["BsmtFullBath"]
    + 0.5 * clean_df["BsmtHalfBath"]
)

print(clean_df[[
    "FullBath",
    "HalfBath",
    "BsmtFullBath",
    "BsmtHalfBath",
    "TotalBathrooms"
]].head())
```

```python
# ============================================
# 33. FEATURE ENGINEERING — ÂGE DU LOGEMENT
# ============================================

current_year = 2026

clean_df["HouseAge"] = (
    current_year - clean_df["YearBuilt"]
)

print(clean_df[[
    "YearBuilt",
    "HouseAge"
]].head())
```

```python
# ============================================
# 34. FEATURE ENGINEERING — ÂGE DEPUIS RÉNOVATION
# ============================================

clean_df["RemodAge"] = (
    current_year - clean_df["YearRemodAdd"]
)

print(clean_df[[
    "YearRemodAdd",
    "RemodAge"
]].head())
```

```python
# ============================================
# 35. VÉRIFICATION DU FEATURE ENGINEERING
# ============================================

new_features = [
    "TotalSF",
    "TotalBathrooms",
    "HouseAge",
    "RemodAge"
]

print(clean_df[new_features].describe())
```
