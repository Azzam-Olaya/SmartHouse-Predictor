```python
# ============================================
# 14. DÉTECTION DES VALEURS EXTRÊMES
# ============================================

# On utilise ici les quartiles pour repérer les valeurs
# qui peuvent être considérées comme extrêmes.

Q1 = df["SalePrice"].quantile(0.25)
Q3 = df["SalePrice"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["SalePrice"] < lower_bound) |
    (df["SalePrice"] > upper_bound)
]

print("Q1 :", Q1)
print("Q3 :", Q3)
print("IQR :", IQR)

print("\nLimite inférieure :", lower_bound)
print("Limite supérieure :", upper_bound)

print("\nNombre de valeurs extrêmes :", len(outliers))
```

```python
# ============================================
# 15. CRÉATION D'UNE COPIE POUR LE NETTOYAGE
# ============================================

clean_df = df.copy()

print("Dataset original :", df.shape)
print("Dataset à nettoyer :", clean_df.shape)
```

```python
# ============================================
# 16. VALEURS MANQUANTES AVANT NETTOYAGE
# ============================================

missing_before = clean_df.isna().sum()

missing_before = (
    missing_before[missing_before > 0]
    .sort_values(ascending=False)
)

print("Nombre de colonnes avec des valeurs manquantes :",
      len(missing_before))

missing_before
```

```python
# ============================================
# VÉRIFICATION APRÈS NETTOYAGE
# ============================================

missing_after = clean_df.isna().sum()

missing_after = (
    missing_after[missing_after > 0]
    .sort_values(ascending=False)
)

print("Nombre de colonnes avec des valeurs manquantes :", len(missing_after))
print("Nombre total de valeurs manquantes :", missing_after.sum())
```

```python
# ============================================
# 17. VARIABLES CATÉGORIELLES
#     où NaN = absence de caractéristique
# ============================================

none_columns = [
    "PoolQC",
    "MiscFeature",
    "Alley",
    "Fence",
    "MasVnrType",
    "FireplaceQu",

    # Garage
    "GarageType",
    "GarageFinish",
    "GarageQual",
    "GarageCond",

    # Basement
    "BsmtQual",
    "BsmtCond",
    "BsmtExposure",
    "BsmtFinType1",
    "BsmtFinType2"
]

for col in none_columns:
    clean_df[col] = clean_df[col].fillna("None")

print("Colonnes traitées :", len(none_columns))
```

```python
# ============================================
# 18. VARIABLES NUMÉRIQUES
#     où NaN correspond à une absence
# ============================================

zero_columns = [
    # Masonry
    "MasVnrArea",

    # Basement
    "BsmtFinSF1",
    "BsmtFinSF2",
    "BsmtUnfSF",
    "TotalBsmtSF",
    "BsmtFullBath",
    "BsmtHalfBath",

    # Garage
    "GarageCars",
    "GarageArea",
    "GarageYrBlt"
]

for col in zero_columns:
    clean_df[col] = clean_df[col].fillna(0)

print("Colonnes traitées :", len(zero_columns))
```

```python
# ============================================
# 19. TRAITEMENT DE LOTFRONTAGE
# ============================================

clean_df["LotFrontage"] = (
    clean_df.groupby("Neighborhood")["LotFrontage"]
    .transform(lambda x: x.fillna(x.median()))
)
```

```python
clean_df["LotFrontage"].fillna(
    clean_df["LotFrontage"].median()
)
```

```python
# ============================================
# 20. VARIABLES CATÉGORIELLES
#     AVEC TRÈS PEU DE VALEURS MANQUANTES
# ============================================

categorical_mode_columns = [
    "MSZoning",
    "Utilities",
    "Functional",
    "Exterior1st",
    "Exterior2nd",
    "Electrical",
    "KitchenQual",
    "SaleType"
]

for col in categorical_mode_columns:
    clean_df[col] = clean_df[col].fillna(
        clean_df[col].mode()[0]
    )

print("Colonnes traitées :", len(categorical_mode_columns))
```
